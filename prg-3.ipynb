```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace(0,5,11)
y=x**2
x
```

```python
y
```

```python
plt.plot(x,y, 'r') #r is the color red
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('X-Y Graph')
plt.show
```

```python
#plt.subplot(nrow,ncols,plot_number)
plt.subplot(1,3,2)
plt.plot(x,y,'b--')
plt.subplot(1,3,1)
plt.plot(x,y,'r*-')
plt.subplot(1,3,3)
plt.plot(x,y,'g^')
```

```python
#using object oriented(subplot is object oriented concept)
#create figure(empty canvas)
fig = plt.figure()
#add set of axes to figure
axes=fig.add_axes([0.5,0.9,0.99,0.89])#left,bottom,widht,heigth
axes.plot(x,y,'b')
axes.set_xlabel('X Axis')
axes.set_ylabel('Y Axis')
axes.set_title('X-Y graph')
plt.Text(0.5 , 1.0 , 'X-Y graph')
```

```python
fig=plt.figure()
axes1=fig.add_axes([0.7,0.6,0.9,0.9])
axes2=fig.add_axes([0.8,0.99,0.4,0.4])
#main figure Axes1
axes1.plot(x,y,'b')
axes1.set_xlabel('X Axis')
axes2.set_ylabel('Y Axis')
axes1.set_title('X-Y Graph')

#insert Figure Axes2
axes2.plot(y,x,'r')#inverted 
axes2.set_xlabel('X Axis')
axes2.set_ylabel('Y Axis')
axes2.set_title('X-Y Graph')
plt.Text(0.5, 1.0, 'X-Y Graph')
```

```python
fig, axes=plt.subplots()
axes.plot(x,y,'b')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('x-y graph')
```

```python
fig,axes=plt.subplots(nrows=1,ncols=2)
plt.Text(0.5, 1.0, 'X-Y Graph')
```

```python
for ax in axes:
    ax.plot(x,y,'b')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('x-y graph')
    fig.tight_layout()
fig    
```

```python
fig=plt.figure(figsize=(8,4), dpi=100)
#<Figure size 800x400 with 0 Axes>
fig,axes=plt.subplots(figsize=(8,3))
axes.plot(x,y,'b')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('x-y graph')
plt.Text(0.5, 1.0, 'x-y graph')
```

```python
fig.savefig('x-y_graph.png',dpi=200)
```

```python
#legend
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='x**2')
ax.plot(x,x**3,label='x**3')
ax.legend(loc=2)
```

```python
ax.legend(loc=1)
ax.legend(loc=2)
ax.legend(loc=3)
ax.legend(loc=4)
ax.legend(loc=0)
```

```python
#line and Marker styles

import matplotlib.pyplot as plt 
import numpy as np

plt.plot(x,x+3,color='b',lw=3,marker='o',markersize=12,linestyle='dashed',markeredgecolor='green',markeredgewidth=4,markerfacecolor='red')
plt.plot(x,x+4,color='r',lw=4,marker='o',markersize=10,linestyle='dashed',markeredgecolor='blue',markeredgewidth=5,markerfacecolor='green')

plt.show()
```

```python
import seaborn as sns
tips=sns.load_dataset("tips")
tips.head()
```

```python
sns.displot(tips["tip"])
```

```python
sns.distplot(tips['total_bill'],kde=False,bins=30)
```

```python
#Jointplot

#kind parameters are 1.scatter , 2.hex , 3.reg , 4.resid , 5.hex
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='scatter')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='hex')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='reg')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='resid')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='kde') 
```
